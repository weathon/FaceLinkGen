```python
from deepface import DeepFace
import os
```

```python
images = os.listdir("./training-tpdne")
```

```python
from PIL import Image
os.makedirs("./tpdne_cropped", exist_ok=True)
for img_name in images:
    p = os.path.join("./training-tpdne", img_name)
    try:
        img_s = DeepFace.extract_faces(p, detector_backend = "opencv", enforce_detection=True)[0]["face"] 
        img_s = Image.fromarray((img_s * 255).astype("uint8"))
        img_s.save(os.path.join("./tpdne_cropped", img_name))
    except Exception as e:
        print(f"Error processing {p}: {e}")
```
