```python
from PIL import Image
import numpy as np
import random


def shuffle_into_64_blocks(image_path, output_path="shuffled.png"):
    img = Image.open(image_path).convert("RGB")
    arr = np.array(img)
    h, w, c = arr.shape

    bh = h // 8
    bw = w // 8

    h_trim = bh * 8
    w_trim = bw * 8
    arr = arr[:h_trim, :w_trim]

    blocks = []
    for y in range(0, h_trim, bh):
        for x in range(0, w_trim, bw):
            blocks.append(arr[y:y+bh, x:x+bw].copy())

    random.shuffle(blocks)

    out = np.zeros_like(arr)
    i = 0
    for y in range(0, h_trim, bh):
        for x in range(0, w_trim, bw):
            out[y:y+bh, x:x+bw] = blocks[i]
            i += 1

    return out

if __name__ == "__main__":
    shuffle_into_64_blocks("gen.png", "output_alex.png")
```
